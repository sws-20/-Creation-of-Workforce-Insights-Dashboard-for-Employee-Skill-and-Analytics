## Data Cleaning and Preprocessing

### Objective

This notebook prepares the raw workforce dataset for downstream analytics and machine learning.

The preprocessing workflow includes:
- Data quality validation
- Identification of non-informative attributes
- Data type validation
- Categorical value standardization
- Feature engineering
- Target variable preparation
- Final dataset validation
- Export of the processed dataset

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv(
    "../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

df.head()
```

```python
df_clean = df.copy()
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
if duplicate_count > 0:
    df_clean = df_clean.drop_duplicates()
```

```python
missing_values = df_clean.isnull().sum()

missing_values[missing_values > 0]
```

```python
constant_columns = [
    column for column in df_clean.columns
    if df_clean[column].nunique() <= 1
]

constant_columns
```

```python
for column in constant_columns:
    print(column)
    print(df_clean[column].unique())
    print()
```

```python
df_clean = df_clean.drop(columns=constant_columns)
```

```python
df_clean["EmployeeNumber"].nunique()
```

```python
len(df_clean)
```

```python
df_clean = df_clean.rename(
    columns={"EmployeeNumber": "EmployeeID"}
)
```

```python
df_clean.head()
```

```python
categorical_columns = df_clean.select_dtypes(
    include="object"
).columns

categorical_columns
```

```python
for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df_clean[column].value_counts())
```

```python
df_clean["AttritionFlag"] = (
    df_clean["Attrition"]
    .map({"No": 0, "Yes": 1})
)
```

```python
df_clean[["Attrition", "AttritionFlag"]].head(10)
```

```python
df_clean["AttritionFlag"].value_counts()
```

```python
df_clean["TenureGroup"] = pd.cut(
    df_clean["YearsAtCompany"],
    bins=[-1, 2, 5, 10, np.inf],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "10+ Years"
    ]
)
```

```python
df_clean["TenureGroup"].value_counts()
```

```python
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[0, 25, 35, 45, 55, np.inf],
    labels=[
        "Under 25",
        "25-35",
        "36-45",
        "46-55",
        "56+"
    ]
)
```

```python
df_clean["AgeGroup"].value_counts()
```

```python
df_clean["YearsSincePromotionRatio"] = (
    df_clean["YearsSinceLastPromotion"]
    / df_clean["YearsAtCompany"].replace(0, np.nan)
)
```

```python
df_clean["YearsSincePromotionRatio"] = (
    df_clean["YearsSincePromotionRatio"].fillna(0)
)
```

```python
df_clean["ExperienceGroup"] = pd.cut(
    df_clean["TotalWorkingYears"],
    bins=[-1, 2, 5, 10, 20, np.inf],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```

```python
df_clean["ExperienceGroup"].value_counts()
```
