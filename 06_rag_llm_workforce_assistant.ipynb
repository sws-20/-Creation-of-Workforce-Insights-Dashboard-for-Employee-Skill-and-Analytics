```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Convert documents into TF-IDF vectors

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

document_vectors = vectorizer.fit_transform(all_documents)

print("Vectorization complete!")
print("Number of documents:", document_vectors.shape[0])
print("Number of features:", document_vectors.shape[1])
```

```python
import faiss
import numpy as np

# Convert sparse TF-IDF matrix to dense float32 matrix
vectors = document_vectors.toarray().astype("float32")

# Normalize vectors for cosine similarity
faiss.normalize_L2(vectors)

# Create FAISS index
index = faiss.IndexFlatIP(vectors.shape[1])

# Add document vectors
index.add(vectors)

print("FAISS index created successfully!")
print("Documents indexed:", index.ntotal)
```

```python
def retrieve_documents(query, top_k=5):
    """
    Retrieve the most relevant workforce documents.
    """

    query_vector = vectorizer.transform([query]).toarray().astype("float32")

    faiss.normalize_L2(query_vector)

    scores, indices = index.search(query_vector, top_k)

    results = []

    for score, idx in zip(scores[0], indices[0]):
        results.append({
            "score": float(score),
            "document": all_documents[idx]
        })

    return results
```

```python
query = "Which department has the highest attrition?"

results = retrieve_documents(query, top_k=5)

for i, result in enumerate(results, 1):
    print("=" * 80)
    print("RESULT", i)
    print("Similarity:", round(result["score"], 4))
    print(result["document"])
```

```python
def get_rag_context(query, top_k=5):
    """
    Retrieve relevant workforce information for a user question.
    """
    
    results = retrieve_documents(query, top_k=top_k)
    
    context_parts = []
    
    for i, result in enumerate(results, 1):
        context_parts.append(
            f"--- Retrieved Information {i} ---\n"
            f"{result['document']}"
        )
    
    return "\n\n".join(context_parts)
```

```python
def create_rag_prompt(query, context):
    
    prompt = f"""
You are an AI Workforce Analytics Assistant.

Your job is to answer questions about the employee workforce dataset.

Use ONLY the information provided in the retrieved context below.

Do not invent facts.
Do not use outside knowledge.
If the retrieved context does not contain enough information to answer the question,
say that the information is not available in the current workforce dataset.

Retrieved Context:
------------------
{context}
------------------

User Question:
{query}

Answer clearly and concisely.
If numerical information is available, include the relevant numbers.

Answer:
"""
    
    return prompt
```

```python
# ============================================================
# COMPLETE RAG RECOVERY CELL
# Rebuilds dataset -> documents -> TF-IDF -> FAISS -> retrieval
# ============================================================

import os
import pandas as pd
import numpy as np
import faiss
from sklearn.feature_extraction.text import TfidfVectorizer

print("=" * 70)
print("REBUILDING RAG RETRIEVAL SYSTEM")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

dataset_path = r"C:\Users\INDIA\OneDrive\Pictures\AI-Powered-Workforce-Analytics\data\processed\workforce_processed.csv"

df = pd.read_csv(dataset_path)

print("\n[1] Dataset loaded")
print("Rows:", len(df))
print("Columns:", len(df.columns))


# ------------------------------------------------------------
# 2. CREATE EMPLOYEE DOCUMENTS
# ------------------------------------------------------------

documents = []

for _, row in df.iterrows():

    text = f"""
Employee ID: {row['EmployeeID']}
Age: {row['Age']}
Age Group: {row['AgeGroup']}
Gender: {row['Gender']}

Department: {row['Department']}
Job Role: {row['JobRole']}
Business Travel: {row['BusinessTravel']}

Daily Rate: {row['DailyRate']}
Hourly Rate: {row['HourlyRate']}
Monthly Income: {row['MonthlyIncome']}
Monthly Rate: {row['MonthlyRate']}
Percent Salary Hike: {row['PercentSalaryHike']}

Education: {row['Education']}
Education Field: {row['EducationField']}

Environment Satisfaction: {row['EnvironmentSatisfaction']}
Job Involvement: {row['JobInvolvement']}
Job Level: {row['JobLevel']}
Job Satisfaction: {row['JobSatisfaction']}
Relationship Satisfaction: {row['RelationshipSatisfaction']}
Performance Rating: {row['PerformanceRating']}

OverTime: {row['OverTime']}
Work Life Balance: {row['WorkLifeBalance']}
Marital Status: {row['MaritalStatus']}
Number of Companies Worked: {row['NumCompaniesWorked']}

Total Working Years: {row['TotalWorkingYears']}
Years At Company: {row['YearsAtCompany']}
Years In Current Role: {row['YearsInCurrentRole']}
Years Since Last Promotion: {row['YearsSinceLastPromotion']}
Years With Current Manager: {row['YearsWithCurrManager']}
Training Times Last Year: {row['TrainingTimesLastYear']}

Attrition: {row['Attrition']}
Attrition Flag: {row['AttritionFlag']}
Tenure Group: {row['TenureGroup']}
Experience Group: {row['ExperienceGroup']}
Years Since Promotion Ratio: {row['YearsSincePromotionRatio']}
"""

    documents.append(text.strip())

print("[2] Employee documents created:", len(documents))


# ------------------------------------------------------------
# 3. CREATE WORKFORCE SUMMARY DOCUMENTS
# ------------------------------------------------------------

summary_documents = []

# Overall workforce summary
overall_attrition = (df["Attrition"] == "Yes").mean() * 100

summary_documents.append(f"""
Workforce Overview:

Total Employees: {len(df)}
Attrition Employees: {(df['Attrition'] == 'Yes').sum()}
Employees Who Stayed: {(df['Attrition'] == 'No').sum()}
Overall Attrition Rate: {overall_attrition:.2f}%

Average Age: {df['Age'].mean():.2f}
Average Job Satisfaction: {df['JobSatisfaction'].mean():.2f}
Average Performance Rating: {df['PerformanceRating'].mean():.2f}
Average Monthly Income: {df['MonthlyIncome'].mean():.2f}
""".strip())


# Department summaries
department_stats = (
    df.groupby("Department")
    .agg(
        Employees=("EmployeeID", "count"),
        Attrition_Count=("AttritionFlag", "sum"),
        Average_Job_Satisfaction=("JobSatisfaction", "mean"),
        Average_Performance=("PerformanceRating", "mean"),
        Average_Income=("MonthlyIncome", "mean")
    )
    .reset_index()
)

for _, row in department_stats.iterrows():

    attrition_rate = (
        row["Attrition_Count"] / row["Employees"] * 100
        if row["Employees"] > 0 else 0
    )

    summary_documents.append(f"""
Department Analysis:

Department: {row['Department']}
Employees: {int(row['Employees'])}
Attrition Count: {int(row['Attrition_Count'])}
Attrition Rate: {attrition_rate:.2f}%

Average Job Satisfaction: {row['Average_Job_Satisfaction']:.2f}
Average Performance Rating: {row['Average_Performance']:.2f}
Average Monthly Income: {row['Average_Income']:.2f}
""".strip())


# ------------------------------------------------------------
# 4. CREATE JOB ROLE SUMMARIES
# ------------------------------------------------------------

jobrole_stats = (
    df.groupby("JobRole")
    .agg(
        Employees=("EmployeeID", "count"),
        Attrition_Count=("AttritionFlag", "sum")
    )
    .reset_index()
)

for _, row in jobrole_stats.iterrows():

    attrition_rate = (
        row["Attrition_Count"] / row["Employees"] * 100
        if row["Employees"] > 0 else 0
    )

    summary_documents.append(f"""
Job Role Analysis:

Job Role: {row['JobRole']}
Employees: {int(row['Employees'])}
Attrition Count: {int(row['Attrition_Count'])}
Attrition Rate: {attrition_rate:.2f}%
""".strip())


# ------------------------------------------------------------
# 5. COMBINE ALL DOCUMENTS
# ------------------------------------------------------------

all_documents = documents + summary_documents

print("[3] Summary documents created:", len(summary_documents))
print("[4] Total RAG documents:", len(all_documents))


# ------------------------------------------------------------
# 6. CREATE TF-IDF VECTORIZER
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

document_vectors = vectorizer.fit_transform(all_documents)

print("[5] TF-IDF vectorization completed")
print("Features:", document_vectors.shape[1])


# ------------------------------------------------------------
# 7. CREATE FAISS INDEX
# ------------------------------------------------------------

vectors = document_vectors.toarray().astype("float32")

faiss.normalize_L2(vectors)

index = faiss.IndexFlatIP(vectors.shape[1])

index.add(vectors)

print("[6] FAISS index created")
print("Indexed documents:", index.ntotal)


# ------------------------------------------------------------
# 8. RETRIEVAL FUNCTION
# ------------------------------------------------------------

def retrieve_documents(query, top_k=5):

    query_lower = query.lower()

    # Special handling for department comparison questions.
    # This makes analytical questions such as
    # "Which department has the highest attrition?"
    # reliable instead of depending only on word similarity.

    if (
        "highest attrition" in query_lower
        or "highest turnover" in query_lower
        or "department attrition" in query_lower
        or "attrition by department" in query_lower
    ):

        department_documents = []

        for doc in summary_documents:
            if "Department Analysis:" in doc:
                department_documents.append(doc)

        # Sort department summaries by actual attrition rate
        def get_attrition_rate(doc):
            for line in doc.splitlines():
                if line.startswith("Attrition Rate:"):
                    return float(
                        line.replace("Attrition Rate:", "")
                            .replace("%", "")
                            .strip()
                    )
            return 0

        department_documents = sorted(
            department_documents,
            key=get_attrition_rate,
            reverse=True
        )

        results = []

        for doc in department_documents[:top_k]:
            results.append({
                "score": get_attrition_rate(doc) / 100,
                "document": doc
            })

        return results


    # Normal semantic TF-IDF + FAISS retrieval
    query_vector = vectorizer.transform(
        [query]
    ).toarray().astype("float32")

    faiss.normalize_L2(query_vector)

    scores, indices = index.search(query_vector, top_k)

    results = []

    for score, idx in zip(scores[0], indices[0]):

        results.append({
            "score": float(score),
            "document": all_documents[idx]
        })

    return results


# ------------------------------------------------------------
# 9. RAG CONTEXT FUNCTION
# ------------------------------------------------------------

def get_rag_context(query, top_k=5):

    results = retrieve_documents(query, top_k)

    context_parts = []

    for i, result in enumerate(results, 1):

        context_parts.append(
            f"--- Retrieved Information {i} ---\n"
            f"{result['document']}"
        )

    return "\n\n".join(context_parts)


# ------------------------------------------------------------
# 10. VERIFY EVERYTHING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RAG SYSTEM READY")
print("=" * 70)

print("df exists:", "df" in globals())
print("documents:", len(documents))
print("summary documents:", len(summary_documents))
print("all_documents:", len(all_documents))
print("vectorizer exists:", "vectorizer" in globals())
print("FAISS index:", index.ntotal)
print("retrieve_documents exists:", "retrieve_documents" in globals())
print("get_rag_context exists:", "get_rag_context" in globals())
```

```python
# ============================================================
# FINAL AI WORKFORCE ASSISTANT
# RAG + RELIABLE WORKFORCE ANALYTICS
# ============================================================

def ask_workforce_assistant(query, top_k=3):

    query_lower = query.lower().strip()

    # --------------------------------------------------------
    # 1. Retrieve relevant RAG context
    # --------------------------------------------------------

    context = get_rag_context(query, top_k=top_k)

    # --------------------------------------------------------
    # 2. Department attrition question
    # --------------------------------------------------------

    if (
        "highest attrition" in query_lower
        or "highest turnover" in query_lower
        or "department has the highest" in query_lower
    ):

        department_stats = (
            df.groupby("Department")
            .agg(
                Employees=("EmployeeID", "count"),
                Attrition_Count=("AttritionFlag", "sum")
            )
            .reset_index()
        )

        department_stats["Attrition_Rate"] = (
            department_stats["Attrition_Count"]
            / department_stats["Employees"]
            * 100
        )

        highest = department_stats.loc[
            department_stats["Attrition_Rate"].idxmax()
        ]

        return (
            f"{highest['Department']} has the highest attrition rate "
            f"at {highest['Attrition_Rate']:.2f}%. "
            f"It has {int(highest['Attrition_Count'])} attrition employees "
            f"out of {int(highest['Employees'])} employees."
        )

    # --------------------------------------------------------
    # 3. Sales / department employee count
    # --------------------------------------------------------

    if (
        "how many employees" in query_lower
        and "sales" in query_lower
    ):

        count = int(
            (df["Department"].str.lower() == "sales").sum()
        )

        return (
            f"The Sales department has {count} employees."
        )

    # --------------------------------------------------------
    # 4. Overall attrition rate
    # --------------------------------------------------------

    if (
        "overall attrition rate" in query_lower
        or "overall attrition" in query_lower
    ):

        attrition_rate = (
            (df["Attrition"] == "Yes").mean() * 100
        )

        return (
            f"The overall workforce attrition rate is "
            f"{attrition_rate:.2f}%."
        )

    # --------------------------------------------------------
    # 5. Department employee count - general
    # --------------------------------------------------------

    departments = df["Department"].dropna().unique()

    for department in departments:

        if department.lower() in query_lower:

            if "employee" in query_lower or "people" in query_lower:

                count = int(
                    (df["Department"] == department).sum()
                )

                return (
                    f"The {department} department has "
                    f"{count} employees."
                )

    # --------------------------------------------------------
    # 6. Fallback to FLAN-T5 for general questions
    # --------------------------------------------------------

    prompt = f"""
Answer the following workforce analytics question using ONLY
the information in the context.

Be concise and give the exact answer.

Context:
{context}

Question:
{query}

Answer:
"""

    response = generator(
        prompt,
        max_new_tokens=80,
        do_sample=False
    )

    generated = response[0]["generated_text"]

    # Remove prompt if the model repeats it
    if "Answer:" in generated:
        generated = generated.split("Answer:")[-1].strip()

    return generated.strip()
```
